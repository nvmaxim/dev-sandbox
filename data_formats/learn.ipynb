```python
import xml.etree.ElementTree as ET

tree = ET.parse("1.xml")

root = tree.getroot()

result = {"store_name": root.find("name").text, "location": root.attrib.get("city"), "staff": []}

for person in root.find("employees").findall("person"):
    result["staff"].append({"name": person.text, "role": person.attrib.get("role")})
print(result)
```

```python
import xml.etree.ElementTree as ET

tree = ET.parse("2.xml")

root = tree.getroot()

result = {"track_name": root.find("Document/name").text, "points": []}

for placemark in root.find("Document").findall("Placemark"):
    timestamp = placemark.find("timestamp").text
    point_teg = placemark.find("Point")

    result["points"].append({"time": timestamp, "lat": float(point_teg.attrib.get("lat")), "lon": float(point_teg.attrib.get("lon"))})

print(result)



```
