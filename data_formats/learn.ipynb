```python
import xml.etree.ElementTree as ET

with open ("2.xml", "r", encoding="utf-8") as file:
    tree= ET.parse(file)
    root = tree.getroot()

    result = {"name_track": root.find("Document/name").text, "points": []}

    for placemark in root.find("Document").findall("Placemark"):
        time = placemark.find("timestamp").text
        point_tag = placemark.find("Point")

        result["points"].append({
            "time": time,
            "lat": float(point_tag.attrib.get("lat")),
            "lon": float(point_tag.attrib.get("lon"))
        })


print(result)
```
