```python
import pandas as pd
import numpy as np

df_result = df.groupby("department").agg({"manager": "count", "revenue": "sum"}).rename(columns={"manager": "manager_count", "revenue": "revenue_total"}).reset_index()
df_result["revenue_share"] = df_result["revenue_total"] / df_result["revenue_total"].sum() 
df_result["avg_per_manager"] = df_result["revenue_total"] / df_result["manager_count"]

df_result["high_perfomance_flag"] = np.where(df_result["avg_per_manager"] > 15000, "Yes", "No")



print(df_result)
```

```python
def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def merge_sort1(xs):
    width = 0
    n = len(xs)
    while width < n:
        for i in range(0, n, 2 * width):
            left = xs[i:i * width]
            right = xs[i * width:i + 2 * width]
            xs[i:i + 2 * width] = merge(left, right)
        width *= 2
    return xs


def  merge_sort2(xs):
    if len(xs) <= 1:
        return xs

    mid = len(xs) // 2
    left = merge_sort2(xs[:mid])
    right = merge_sort2(xs[mid:])
    return merge(left, right)
```
