```python

def is_prime(n):
    try:
        n = int(float(n))
    except (TypeError, ValueError):
        return False
    if n < 2:
        return False
    for i in range(0, int(n ** 0,5) + 1):
        if n % i == 0:
            return False
    return True


print(is_prime(2.0))
print(is_prime('2017'))
print(is_prime('12709189.00'))
print(is_prime(1))
print(is_prime('abc'))
```
