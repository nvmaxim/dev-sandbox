```python
import inspect

def inspect_function(func):
    print(f"имя функции - {func.__name__}")

    # print(f"документация - {func.__doc__}")
    doc = inspect.getdoc(func)
    print(f"документация - {doc if doc else 'нет докуменатции'}")

    params = inspect.signature(func).parameters

    positional_only = []
    positional_or_keyword = []
    keyword_only = []
    var_args, var_kwargs = None, None
    defaults, default_kwargs = [], []

    for name, param in params.items():
        if param.kind == param.POSITIONAL_ONLY:
            positional_only.append(name)
        elif param.kind == param.POSITIONAL_OR_KEYWORD:
            positional_or_keyword.append(name)
            if param.default != param.empty:
                defaults.append(f"{name} - {param.default}")
        elif param.kind == param.KEYWORD_ONLY:
            keyword_only.append(name)
            if param.default != param.empty:
                default_kwargs.append(f"{name} - {param.default}")
        elif param.kind == param.VAR_POSITIONAL:
            var_args = name
        elif param.kind == param.VAR_KEYWORD:
            var_kwargs = name


    print(f"аргументы принимаемые функцией {", ".join(params.keys())}")
    print(f"строго позиционные {", ".join(positional_only)}")
    print(f"поизиционные или именованные {", ".join(positional_or_keyword)}")
    print(f"строго именновые {", ".join(keyword_only)}")
    print(f"имя для аргс {var_args if var_args else ""}")
    print(f"имя для kwargs {var_kwargs if var_kwargs else ""}")
    print(f"дефолтные значения {", ".join(defaults)}")
    print(f"дефолтные значения {", ".join(default_kwargs)}")

    pass



# tests for foo function для тестирования
# 1. Создаем тестовую функцию со всеми типами аргументов
def my_sample_function(a, b=42, *args, c, d=100, **kwargs):
    """Это тестовая функция для демонстрации работы inspect_function. Она принимает позиционные, именованные аргументы, args и kwargs."""


    pass


# 2. Вызываем нашу функцию инспектирования, передавая ей my_sample_function
inspect_function(my_sample_function)
```
