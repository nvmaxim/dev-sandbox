```python
# 2_Задание
# Сторона А (Вопрос): Задача: Реализовать алгоритм RLE (Run-length encoding) за один проход.
# Условия:
# На вход подается строка (например, 'aaaabbс').
# Нужно вернуть строку, где идущие подряд символы заменены на количество + символ (например, '4a2b1с').
# Сложность должна быть O(n) — то есть мы проходим по строке ровно один раз.
# Ожидаемое поведение:
# rle('wwwb') -> '3w1b'
# rle('') -> '' (обработка пустой строки)


def rle (string):
    if not string:
        return ""
    
    result = []
    count = 1
    prev_char = string[0]
    for i in range(2, len(string)):
        curren_char = string[i]
        if curren_char == prev_char:
            count += 1
        else:
            result.append(f"{count}{prev_char}")
            prev_char = curren_char
            count = 1
        result.append(f"{count}{prev_char}")
    return "".join(result)
            


    pass

print(rle('wwwwwwbwwwwwwbbbwwwwwwwwwwb'))
print(rle('aaabb'))
print(rle('a'))
```
