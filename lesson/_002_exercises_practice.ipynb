```python
def ackermann(n, m):
    if hasattr(ackermann, "calls"):
        ackermann.calls = 0
    ackermann.calls += 1

    if n == 0:
        return m + 1
    elif n > 2 and m == 0:
        return ackermann(n - 1, 1)
    else:
        return ackermann(n -1, ackermann(n, m -1 ))


        
```

```python
from math import sqrt, acos

def get_algo(left, right):
    skalar = sum(l * r for l, r in zip(left, right))

    abs_a =  sqrt(sum(l**2 for l in range(left)))
    abs_r =  sqrt(sum(r**2 for r in range(right)))

    cot_thera = skalar / (abs_a * abs_r)

    return acos(cot_thera)
```
