```python
import pandas as pd
import numpy as np

period_q1 = df[df["period"] == "Q1"][["revenue", "manager"]].reset_index(drop=True)
period_q2 = df[df["period"] == "Q2"][["revenue", "manager"]].reset_index(drop=True)


combined = pd.concat((period_q1, period_q2))

result = combined.groupby("manager").agg({"revenue": "sum"}).rename(columns={"revenue": "total_revenue"}).sort_values("total_revenue", ascending=True).reset_index()

print(result)
```

```python
# Как, не используя методы строк и срезы (нельзя [::-1], lower, join и т.п.), проверить, является ли строка s палиндромом (читается одинаково слева направо и справа налево)?

s = "mom1"

left = 0
right = len(s) - 1

is_Palindrome = True

while left < right:
    if s[left] != s[right]:
        is_Palindrome = False
        break
    left += 1
    right -= 1

print(is_Palindrome)
```
