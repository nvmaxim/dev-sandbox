```python
def matrix(n):
    mat = [[0]* n for _ in range(n)]

    direction = [()]
    curr_dir = 0
    y, x = 0, 0
    
    for num in range(1, n*n + 1):
        mat[y][x]= num
        dy = direction[curr_dir]
        dy, dx = curr_dir  
        new_y, new_x = dy+y, dx+ x

        if not (0 < new_y <= n and 0 < new_x <= n and mat[new_y][new_x] == 0):
            curr_dir = (curr_dir + 1) % 4
            dy, dx = curr_dir  
            new_y, new_x = dy+y, dx+ x

        y, x = new_y, new_x
    return mat
```
