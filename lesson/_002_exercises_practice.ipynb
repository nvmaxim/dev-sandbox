```python
from _001_main import df_csv
import pandas as pd
import numpy as np
df = df_csv

print(df)
```

```python
import pandas as pd
import numpy as np

period_q1 = df[df["period"] == "Q1"][["revenue", "manager"]].reset_index(drop=True)
period_q2 = df[df["period"] == "Q2"][["revenue", "manager"]].reset_index(drop=True)


combined = pd.concat((period_q1, period_q2))

result = combined.groupby("manager").agg({"revenue": "sum"}).rename(columns={"revenue": "total_revenue"}).sort_values("total_revenue", ascending=True).reset_index()

print(result)
```

```python
# УПРАЖНЕНИЕ 1: "Анализ эффективности менеджеров (python)
# В каком отделе (А или Б) менеджеры в среднем приносят больше выручки?

result = df[df["department"].isin(("А", "Б"))].groupby("department").agg({"revenue": "sum"}).rename(columns=({"revenue": "total_revenue"})).sort_values("total_revenue", ascending=True).reset_index()

print(result)
```
