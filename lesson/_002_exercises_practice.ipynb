```python
from _001_main import df_csv
import pandas as pd
import numpy as np
df = df_csv

print(df)
```

```python
# УПРАЖНЕНИЕ 5: "Комплексный отчет" (python)
# Создайте комплексный отчет по отделам: количество менеджеров, общая выручка, доля в общей выручке, средняя выручка на менеджера и флаг высокой эффективности.
import numpy as np

df = df.groupby("department").agg(manager_count=("manager", "count"), revenue_total=("revenue", "sum"))


df["share_revenue"] = df["revenue_total"] / df["revenue_total"].sum()

df["avg_per_manager"] = df["revenue_total"] / df["manager_count"]

df["high_perfomance"] = np.where(df["avg_per_manager"] > 12000, "Yes", "No")
print(df)
```
