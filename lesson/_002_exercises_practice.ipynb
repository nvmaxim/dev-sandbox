```python
# 1_Задача: Проверка числа на простоту (is_prime) с валидацией типов.
# Условия:
# Реализовать функцию is_prime(n).
# Аргумент n может быть не только int, но и float, и str (например, '2017', '12709189.00').
# Алгоритм должен быть эффективным (подходить для больших чисел, как 18375703).
# Критерии простого числа: строго целое, >1, имеет ровно 2 делителя.
# Ожидаемое поведение:
# is_prime(2.0) -> True
# is_prime('2017') -> True
# is_prime(1) -> False
# is_prime('abc') -> False


def is_prime(n):
    try:
        n = int(float(n))
    except (TypeError, ValueError):
        return False
    
    if n < 2:
        return False
    
    for i in range(2, (n ** 0,5) + 1):
        if n % i == 0:
            return False
    return True


print(is_prime(2.0))
print(is_prime('2017'))
print(is_prime('12709189.00'))
print(is_prime(1))
print(is_prime('abc'))
```

```python
# 3.2_Задание: Векторная математика (Угол между векторами)
# Сторона А (Вопрос): Задача: Найти угол между двумя n-мерными векторами в радианах.
# Указания:
# Векторы — это списки left и right.
# Использовать формулу:​
# Где a⋅b — скалярное произведение,
#  а ∣a∣ — длина (норма) вектора.
from math import sqrt, acos

def vector_angle(l, r):
    dot_val = sum((l * r) for l, r in zip(l, r))
    len_a = sqrt(sum(l**2 for l in l))
    len_b = sqrt(sum(l**2 for r in r))

    cos_t = dot_val / (len_a * len_b)

    return acos(cos_t)  



print(f"Тест 1 (45°): {vector_angle([1, 0, 1], [0, 0, 1]):.5f}")
print(f"Тест 2 (90°): {vector_angle([1, 0], [0, 1]):.5f}")
print(f"Тест 3 (0°):  {vector_angle([5, 5], [5, 5]):.5f}")


```

```python
# Условие: На вход функции matrix подаётся число N. Функция возвращает матрицу-спираль, закрученную против часовой стрелки, размера NxN. Пример для 3х3:


def matrix (n):
    mat = [[0] * n for _ in range(n)]

    directions = [(1, 0), (0, 1), (-1, 0), (0, -1)]

    curr_dir = 0

    y, x = 0, 0

    for num in range(1, n * n + 1):
        mat[y][x] = num
        dy, dx = directions[curr_dir]
        new_y, new_x = y + dy, x + dx

        if not (0 <= new_y < n and 0 <= new_x < n and mat[new_y][new_y] == 0):
            curr_dir = (curr_dir + 1) % 4
            dy, dx = directions[curr_dir]
            new_y, new_x = y + dy, x + dx
        

        y, x = new_y, new_x
    return mat
```
